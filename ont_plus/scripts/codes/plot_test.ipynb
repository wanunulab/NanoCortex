```python
import forgi.graph.bulge_graph as fgb
import forgi.visual.mplot as fvm
import matplotlib.pyplot as plt

def plot_rna_from_ct(ct_filepath, output_filename="rna_structure.png"):
    """
    从 CT 文件绘制 RNA 二级结构图（修正版）。
    """
    try:
        # 修正加载方式：直接使用 fgb.BulgeGraph.from_ct_file
        # 这是 forgi 处理 .ct 文件最底层且最稳定的方法
        bg = fgb.BulgeGraph.from_ct_file(ct_filepath)

        # 初始化绘图
        plt.figure(figsize=(12, 12)) 

        # 使用 forgi.visual.mplot (fvm) 进行绘制
        # 它可以产生非常经典的 RNA 骨架图
        fvm.plot_rna(bg, 
                     text_kwargs={"fontsize": 7, "color": "#333333", "weight": "bold"}, 
                     base_outline_kwargs={"edgecolor": "black", "linewidth": 0.5},
                     column_color_labels=None) # 如果想根据结构上色，可以自定义这个参数

        plt.title(f"RNA Structure: {ct_filepath.split('/')[-1]}", fontsize=15)
        plt.axis("off")

        # 保存并导出
        plt.savefig(output_filename, dpi=300, bbox_inches='tight')
        print(f"✅ 成功！图片已保存至: {output_filename}")
        plt.show()

    except Exception as e:
        print(f"❌ 绘制失败，错误原因: {e}")

if __name__ == "__main__":
    # 请确保路径正确
    ct_file = "/projects/wanunulab/qinixia/projects/ai_agent/nanoporePlus/deps/RNA-FM/redevelop/results/pred_ct/1p6v_B.ct"
    out_file = "/projects/wanunulab/qinixia/projects/ai_agent/nanoporePlus/deps/RNA-FM/redevelop/results/1p6v_B_rna_structure.png"
    
    plot_rna_from_ct(ct_file, out_file)
```
